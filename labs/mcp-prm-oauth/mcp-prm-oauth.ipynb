# APIM ❤️ MCP

## MCP Protected Resources Metadata - PRM

![flow](../../images/mcp-prm-oauth.gif)

Playground to experiment the [Model Context Protocol](https://modelcontextprotocol.io/) with the [client authorization flow](https://modelcontextprotocol.io/specification/2025-03-26/basic/authorization#2-10-third-party-authorization-flow). 

In this flow, Azure API Management act both as an OAuth client connecting to the [Microsoft Entra ID](https://learn.microsoft.com/en-us/entra/architecture/auth-oauth2) authorization server and as an OAuth authorization server for the MCP client ([MCP inspector](https://modelcontextprotocol.io/docs/tools/inspector) in this lab), VS Code or Copilot Studio.

Full documentation of the assets created in this lab and how they interact is documented in [Arch decisions summary](ARCH_SUMMARY.md) and in [MCP PRM Server implementation](src/mcp-prm-server/README.md)


⚠️ This lab implements the [MCP Authorization proposal](https://modelcontextprotocol.io/specification/2025-03-26/basic/authorization) and [RFC9729](https://datatracker.ietf.org/doc/html/rfc9728), this is as close to production-grade security for MCP servers as it gets, and is inspired by the amazing work by [blackchoey/remote-mcp-apim-oauth-prm](https://github.com/blackchoey/remote-mcp-apim-oauth-prm) - originally written in C#

### 🧪 Test authorization **WITH** a token (PRM → Entra → MCP)

This cell mirrors what an MCP client does: fetch **Protected Resource Metadata (PRM)** from APIM, discover the **authorization server** metadata at Entra, then call the MCP endpoint on APIM with a **Bearer access token** you paste into the code cell (for example from MCP Inspector or another OAuth client using the scope in PRM).

APIM should **not** return **401** when the token audience matches the MCP app (entra_app_client_id). The MCP server may return a non-200 status for the test request; passing APIM JWT validation is the primary success criterion here. Full MCP sessions are covered in the VS Code and Inspector sections below.

```python
$MCP_APP_ID = "5d8bf940-156a-4ea0-af2f-6dc7f8c8a4ef"
az login --tenant "293882e2-ff46-4747-b46a-a84ca8e98fa0" --scope "$MCP_APP_ID/user_impersonate"
az account get-access-token --scope "$MCP_APP_ID/user_impersonate" --query accessToken -o tsv
```

```python
# PRM → Entra metadata → pasted token → authenticated call to APIM /mcp
import json
import base64
import os
import requests

# Paste access token from Entra (MCP Inspector, authorization code flow, etc.).
# Scope must match PRM scopes_supported (e.g. {mcpAppId}/user_impersonate). Do not commit tokens to git.
manual_access_token = "xx"
def decode_jwt_payload(token):
    """Decode JWT payload (no signature verification) for display."""
    segment = token.split(".")[1]
    padded = segment + "=" * (-len(segment) % 4)
    return json.loads(base64.urlsafe_b64decode(padded).decode("utf-8"))


access_token = (manual_access_token or "").strip()
if not access_token:
    utils.print_error(
        "Set manual_access_token to your Bearer access token, then re-run this cell."
    )
else:
    # 1) Fetch Protected Resource Metadata (PRM) from APIM
    utils.print_info(f"Fetching PRM from {mcp_prm_endpoint}")
    prm_response = requests.get(mcp_prm_endpoint, timeout=30)
    if prm_response.status_code != 200:
        utils.print_error(f"PRM request failed: HTTP {prm_response.status_code}")
        utils.print_message(prm_response.text)
    else:
        prm = prm_response.json()
        utils.print_ok("Protected Resource Metadata retrieved")
        utils.print_message(json.dumps(prm, indent=2))
        authorization_servers = prm.get("authorization_servers") or []
        scopes_supported = prm.get("scopes_supported") or []
        if not authorization_servers or not scopes_supported:
            raise ValueError("PRM is missing authorization_servers or scopes_supported")
        utils.print_info(f"Expected token scope (from PRM): {scopes_supported[0]}")

        # 2) Fetch authorization server metadata (OpenID Connect discovery)
        auth_server = authorization_servers[0].rstrip("/")
        openid_config_url = f"{auth_server}/.well-known/openid-configuration"
        utils.print_info(f"Fetching authorization server metadata from {openid_config_url}")
        oidc_response = requests.get(openid_config_url, timeout=30)
        if oidc_response.status_code != 200:
            utils.print_error(f"OpenID configuration request failed: HTTP {oidc_response.status_code}")
            utils.print_message(oidc_response.text)
        else:
            oidc = oidc_response.json()
            utils.print_ok("Authorization server metadata retrieved")
            utils.print_message(
                json.dumps(
                    {
                        "issuer": oidc.get("issuer"),
                        "authorization_endpoint": oidc.get("authorization_endpoint"),
                        "token_endpoint": oidc.get("token_endpoint"),
                    },
                    indent=2,
                )
            )

        claims = decode_jwt_payload(access_token)
        utils.print_info(
            f"Token aud={claims.get('aud')} scp={claims.get('scp')} (expected aud={entra_app_client_id})"
        )

        # 3) Call MCP on APIM with Bearer token (streamable HTTP initialize)
        mcp_server_url = f"{apim_resource_gateway_url}/mcp"
        init_body = {
            "jsonrpc": "2.0",
            "id": 1,
            "method": "initialize",
            "params": {
                "protocolVersion": "2024-11-05",
                "capabilities": {},
                "clientInfo": {"name": "lab-notebook", "version": "1.0"},
            },
        }
        utils.print_info(f"Calling MCP endpoint with Bearer token: {mcp_server_url}")
        auth_response = requests.post(
            mcp_server_url,
            headers={
                "Content-Type": "application/json",
                "Accept": "application/json, text/event-stream",
                "Authorization": f"Bearer {access_token}",
            },
            json=init_body,
            timeout=60,
        )

        if auth_response.status_code == 401:
            utils.print_error(
                "Received 401 — check token audience, expiry, and APIM validate-azure-ad-token policy"
            )
            utils.print_message(auth_response.text)
        elif auth_response.status_code == 200:
            utils.print_ok("Received HTTP 200 — APIM accepted the token and the request succeeded")
            utils.print_message(auth_response.text)
        else:
            utils.print_ok(
                f"APIM did not return 401 (HTTP {auth_response.status_code}) — JWT validation passed at the gateway"
            )
            utils.print_info(
                "Non-200 responses can be normal if the backend MCP server errors on this test payload."
            )
            if auth_response.text:
                utils.print_message(auth_response.text[:2000])
```
